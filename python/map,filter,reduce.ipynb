```python
from functools import *
l=[1,5,6,9,2,3]
x=list(map(lambda i:i*i,l))
print(x)
z=list(filter(lambda i: i%2==1,l))
print(z)
y=reduce(lambda i, y:i*y,l)
print(y)
```
